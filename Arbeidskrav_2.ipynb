<span style="color: blue;"><b>Oppgave 3)</b></span>

Lag et program med en funksjon som regner om fra grader til radianer.
Programmet skal starte med:

import numpy as np

v_grad = float(input('Skriv inn gradtallet:' ))

Radiantallet til vinkelen regnes så ut ved følgende formel: 

v_rad = v_grad*np.pi/180

Resultatet v_rad skrives til skjerm med passende tekst og verdi.

Merk: np.pi er en ferdiglaget funksjon som gir verdien 3.1415...

```python
"""
Oppgave 3
Grader og radianer er to ulike måleenheter for vinkler. 
"""

import numpy as np

v_grad = float(input('Skriv inn gradtallet: ' ))  

v_rad = v_grad * np.pi / 180 

print(f"Vinkelen i radianer er: {v_rad:.4f}")   # Skriver ut radiantallet med fire desimaler (np.pi er en ferdiglaget funksjon som gir verdien 3.1415)
```

a) Opprett en dictionary som gitt under. Dictionaryen har ulike land som nøkkel (Keys) og gir info om hovedstaden i landet og antall innbyggere i mill. i hovedstaden.

Bilde fra oppgaveteksten

b) Lag et program som ber brukeren skrive inn et land (eksempelvis England).
Programmet skal på bakgrunn av dette skrive ut følgende setning:

London er hovedstaden i England og det er 8.982 mill. innbyggere i London

c) Lag et program som ber brukeren skrive inn info om et nytt land (altså et land som
ikke allerede finnes i dictionaryen data). Videre skal brukeren oppgi hovedstad og
antall innbyggere for det «nye» landet. Programmet skal så utvide/oppdatere dictionaryen med den nye informasjonen. 
Dictionaryen data skrives så til skjerm.


```python
"""
Oppgave 4a)
Dictionary med land som nøkler (keys). Hvert land har en liste med hovedstad og innbyggertall i millioner.
Hovedstaden har indeks 0, og innbyggertallet har indeks 1.
"""

data = {
    "Norge": ["Oslo", 0.634],
    "England": ["London", 8.982],
    "Frankrike": ["Paris", 2.161],
    "Italia": ["Roma", 2.873]
    }

print(data)
```

```python
"""
Oppgave 4b)
Programmet spør brukeren om et land og finner hovedstaden og antall innbyggere fra dictionaryen.
Land er nøkler (keys), mens hovedstaden har indeks 0 og innbyggertallet har indeks 1.
"""

land = input("Velg et av disse landene (Norge, England, Frankrike eller Italia): ")  

hovedstad = data[land][0]   # Henter hovedstad fra indeks 0 
innbyggertall= data[land][1]   # Henter innbyggertall fra indeks 1

print(f"{hovedstad} er hovedstaden i {land} og det er {innbyggertall} mill. innbyggere i {hovedstad}")
```

```python
"""
Oppgave 4c)
Programmet spør brukeren om et nytt land, hovedstad og innbyggertall.
Opplysningene legges til i dictionaryen med landet som nøkkel (key).
Hovedstad og innbyggertall lagres i en liste med indeks 0 og 1.
"""

land = input("Skriv inn et nytt land:")

hovedstad = input("Skriv inn hovedstaden:")

innbyggertall = float(input("Skriv inn innbyggertallet i millioner:"))

data[land] = [hovedstad, innbyggertall]   # Legger til det nye landet i dictionaryen

print(data)   # Skriver ut den oppdaterte dictionaryen
```

<span style="color: blue;"><b>Oppgave 5)</b></span>

Lag et program med en funksjon som tar a og b som inn-argumenter og som så
regner ut arealet og «ytre» omkrets til en figur satt sammen av en rettvinklet trekant og en
halvsirkel, se figuren under. 

Bilde fra oppgaveteksten

Med «ytre» omkrets menes samlet lengde av de sorte strekene = begge lang sidene + halv sirkel

Funksjonen skal returnere arealet og «ytre» omkrets, som så skrives til skjerm med passende
tekst

```python
"""
Oppgave 5
Regner ut areal og "ytre" omkrets av en rettvinklet trekant og en halvsirkel.
"""

# Den rettvinklede trekanten
a = float(input("Skriv inn lengden a: "))   # Kort katet og diameter
b = float(input("Skriv inn lengden b: "))   # Lang katet

c = (a**2 + b**2)**0.5                      # Bruker Pytagoras' læresetning for å beregne hypotenusen (c)
r = a / 2                                   # Radius 
π = 3.14                                    # π med to desimaler

# Formlene for sirkelens areal og omkrets
areal_sirkel = π * r**2
omkrets_sirkel = 2 * π * r

# Trekantens areal er g*h/2. g tilsvarer a og h tilsvarer b
areal_trekant = a * b / 2
omkrets_trekant = a + b + c 

# Deler sirkelens areal og omkrets på 2, og trekker fra side a
areal = areal_trekant + areal_sirkel / 2
omkrets = omkrets_trekant - a + omkrets_sirkel / 2

print(f"Areal av hele figuren er {areal:.2f}")

print(f"Den ytre omkretsen av hele figuren er {omkrets:.2f}") 
```

<span style="color: blue;"><b>Oppgave 6)</b></span>

Skriv en kode som plotter funksjonen 𝑓(𝑥) = −𝑥**2 − 5, for x på intervallet [-10,10].

Hint: np.linspace(-10, 10, 200) gir en array med 200 punkter jevnt fordelt på intervallet
[-10,10].

```python
"""
oppgave 6
Programmet tegner opp grafen til funksjonen f(x) = -x**2 - 5 for x-verdier fra [-10 til 10].
"""

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 200)   # Lager 200 punkter mellom -10 og 10
y = -x**2 - 5                   # Funksjonen f(x)= -x**2 - 5

plt.plot(x,y)                   # Tegner grafen
plt.xlabel("x")                 # Navn på x-aksen
plt.ylabel("f(x)")              # Navn på y-aksen
plt.title("f(x) = -x**2 -5")    # Overskrift på grafen
plt.grid()                      # Viser rutenett
plt.show()                      # Viser grafen
```
